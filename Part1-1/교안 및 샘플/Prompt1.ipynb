```python
# LangChain에서 실행할 수 있는 모든 대상은 Runnable 객체
# Runnable 객체는 Runnable 인터페이스를 구현한 클래스의 인스턴스

# LangChain에서 invoke, stream, batch는 모두 Runnable 객체를 실행하는 메서드

# invoke는 하나의 입력에 대해 모델을 한 번 호출하여 최종 결과물을 즉시 반환하는 동기(synchronous) 방식의 메서드
# batch는 여러 개의 입력을 리스트 형태로 받아 한 번의 호출로 병렬 처리하고, 각각의 결과물을 리스트로 반환하는 메서드
# stream은 모델의 응답을 실시간으로, 토큰(단어) 단위로 스트리밍하여 반환하는 메서드 (기다리지 않음)
```

```python
# 가장 단순한 프롬프트 예제

# langchain 라이브러리에서 필요한 클래스 임포트
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
#from langchain.chains import LLMChain

# 1. 모델 생성
# ChatGoogleGenerativeAI는 LangChain 클래스
# Google의 생성형 AI 모델을 사용하기 위한 클래스
# ChatGoogleGenerativeAI는 대화 중심, Gemini 모델 지원
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

# 2. PromptTemplate 생성
# {topic} 변수를 포함한 프롬프트 템플릿을 정의
template = "다음 주제에 대해 간결하게 설명해 줘: {topic}"
prompt = PromptTemplate.from_template(template)

# 3. Chain 생성
# 프롬프트와 언어 모델을 연결하여 체인을 생성
# 체인의 요소는 Runnable 객체로서 invoke(), batch(), stream()으로 실행
# Runnable 객체를 체인으로 묶으면 체인 그 자체도 Runnable 객체

chain = prompt | llm

# 4. 체인 실행 및 결과 출력
topic = "LangChain의 프롬프트"

print('---', prompt.format(topic=topic)) # {topic}의 값이 채워진 prompt의 문자열을 확인
response = chain.invoke(topic)

print(response)
```

```python
# 일반 함수를 Runnable 객체로 변환해서 입력 내용 확인하기

# langchain 라이브러리에서 필요한 클래스 임포트
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
# RunnableLambda 클래스 임포트
from langchain_core.runnables import RunnableLambda

# 1. 모델 생성
# ChatGoogleGenerativeAI는 LangChain 클래스
# Google의 생성형 AI 모델을 사용하기 위한 클래스
# ChatGoogleGenerativeAI는 대화 중심, Gemini 모델 지원
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

# 2. PromptTemplate 생성
# {topic} 변수를 포함한 프롬프트 템플릿을 정의
# {topic} 변수의 값은 체인을 호출할 때 전달
template = "다음 주제에 대해 간결하게 설명해 줘: {topic}"
prompt = PromptTemplate.from_template(template)

# 3. Chain 생성
# 프롬프트와 언어 모델을 연결하여 체인을 생성
# 체인의 요소는 Runnable 객체로서 invoke(), batch(), stream()으로 실행
# Runnable 객체를 체인으로 묶으면 체인 그 자체도 Runnable 객체


def log_and_pass(x):
    print(f"\n[로그] 현재 단계 입력값: {x}\n")
    return x
# RunnableLambda은 일반 함수를 Runnable 객체로 변환
# 즉, 일반 함수를 체인에서 실행할 수 있게 하는 역할
# Runnable 객체로 변환된 함수는 이전 단계의 출력을 입력으로 받음
logger = RunnableLambda(log_and_pass)


# 체인에 로거를 추가
# prompt의 입력은 'topic' 변수의 값
# llm의 입력은 prompt의 출력
chain = logger | prompt | logger | llm

# #################### 추가 코드 끝 ####################

# 4. 체인 실행 및 결과 출력
topic = "LangChain의 프롬프트"
response = chain.invoke(topic)

print(response)
```

```python
# 체인의 어떤 단계에서 출력이 둘 이상인 경우 딕셔너리 형태로 만들어야 함

# 1. RunnableParallel
# 2. RunnablePassthrough.assign()
# 3. JsonOutputParser
```

```python
# RunnableParallel

from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableParallel
from langchain_google_genai import ChatGoogleGenerativeAI

# 1. 모델 생성
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

# 2. PromptTemplate 생성
name_prompt = PromptTemplate.from_template(
    "다음 제품 타입에 어울리는 창의적인 이름을 하나만 만들어 주세요:\n\n제품 타입: {product_type}"
)
slogan_prompt = PromptTemplate.from_template(
    "다음 키워드를 사용해 기억에 남는 슬로건을 하나만 만들어 주세요:\n\n키워드: {keyword}"
)

# 3. Chain 생성
name_chain = name_prompt | llm
slogan_chain = slogan_prompt | llm

# 3. RunnableParallel로 두 체인을 병렬 실행
combined_chain = RunnableParallel(
    name=name_chain, # name_prompt | llm
    slogan=slogan_chain # slogan_prompt | llm
)
# combined_chain을 실행하면 다음과 같은 딕셔너리가 리턴

# {
#   "name": "name_chain의 실행 결과",
#   "slogan": "slogan_chain의 실행 결과"
# }

# 4. 최종 문장을 생성하는 프롬프트와 체인
final_prompt = PromptTemplate.from_template(
    "'{name}'이라는 제품을 위한 슬로건은 '{slogan}'입니다."
)

# 5. 모든 체인 결합
final_chain = combined_chain | final_prompt | llm

result = final_chain.invoke({"product_type": "스마트 커피 머신", "keyword": "편리함"})
# {"product_type": "스마트 커피 머신", "keyword": "편리함"} >>> 'name': name_chain(name_prompt | llm) 
# {"product_type": "스마트 커피 머신", "keyword": "편리함"} >>> 'slogan': slogan_chain(slogan_prompt | llm)
print(result.content)
```

```python
# RunnablePassthrough.assign()
# 출력에 키 값을 부여


from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_google_genai import ChatGoogleGenerativeAI

# 1. 모델 생성
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

# 2. '재료'를 생성하는 체인
ingredients_chain = (
    PromptTemplate.from_template("{dish}에 들어가는 재료 3가지를 알려줘.")
    | llm
)

# 3. 최종 프롬프트는 'dish'와 'ingredients'라는 두 개의 변수를 필요
final_prompt = PromptTemplate.from_template(
    "'{dish}'을 만드는 주요 재료는 다음과 같습니다:\n{ingredients}"
)

# 4. RunnablePassthrough.assign()을 사용하여 체인 구성
chain = (
    RunnablePassthrough.assign(
        ingredients=ingredients_chain # PromptTemplate.from_template("{dish}에 들어가는 재료 3가지를 알려줘.") | llm
        # ingredients_chain의 결과에 키 'ingredients'를 부여
        # 체인의 다음 단계인 final_prompt에서 {ingredients}을 사용하기 때문에 거기에 맞춘 것
    )
    | final_prompt
    | llm
)

# 5. 체인 실행
# 초기 입력으로 {"dish": "김치찌개"}를 전달, 이 값은 체인의 모든 단계에 전달
result = chain.invoke({"dish": "김치찌개"})

print(result.content)
```

```python
# PromptTemplate와 PromptTemplate.from_template

from langchain_core.prompts import PromptTemplate

# 1. PromptTemplate
# input_variables를 명시적으로 할당해서 검증
# 검증이란
# 템플릿에 {input}이 있는데 input_variables에 "input"이 있는지 없는지
# input_variables에 "input"이 있는데 템플릿에 {input}이 있는지 없는지

manual_prompt = PromptTemplate(
    input_variables=["animal", "food"],
    template="내가 좋아하는 동물은 {animal}이고, 좋아하는 음식은 {food}야.",
)

# 2. PromptTemplate.from_template
# template 문자열만 제공, 변수를 자동으로 파싱 파싱
auto_prompt = PromptTemplate.from_template(
    "내가 좋아하는 동물은 {animal}이고, 좋아하는 음식은 {food}야."
)

# 두 템플릿은 동일
print(manual_prompt.format(animal="강아지", food="사료"))
print(auto_prompt.format(animal="고양이", food="생선"))
```
